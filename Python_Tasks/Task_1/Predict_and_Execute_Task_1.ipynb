## Q35

**Prediction (before execution):**

```text
[[1, 99], [3, 4]]
```

````python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
````

**Why:** A shallow copy duplicates only the outer list. The inner lists are shared, so changing an inner element through `b` also changes `a`.

**Verification reference:**

```text
[[1, 99], [3, 4]]
```

## Q36

````python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
````

**Why:** A deep copy recursively copies nested lists. Mutating an inner list in `b` does not affect `a`.

**Verification reference:**

```text
[[1, 2], [3, 4]]
```
