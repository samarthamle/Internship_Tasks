## Q1. Find the Shortest Word in a Sentence
**Observe:** Input is a sentence (string); output is one word. Example: `Python makes coding easy` → `easy`.

**Understand:** Return the word with the fewest characters.

**Break:**
1. Split the sentence into words.
2. Assume the first word is the shortest.
3. Compare every other word's length.
4. Update when a shorter one is found.

**Logic:**
```
words = split(sentence)
shortest = words[0]
for each word: if len(word) < len(shortest): shortest = word
return shortest
```


````python
def shortest_word(sentence):
    words = sentence.split()
    if not words:                      # edge case: empty input
        return None
    shortest = words[0]
    for w in words[1:]:
        if len(w) < len(shortest):     # strict '<' keeps the first on ties
            shortest = w
    return shortest

print(shortest_word("Python makes coding easy"))   # easy
print(shortest_word("I love programming"))          # I
print(shortest_word(""))                            # None
````

## Q2. Find the Largest Number in a List (no max())
**Observe:** Input: list of numbers; output: one number. `[12, 45, 7, 89, 23]` → `89`.

**Understand:** Find the biggest value using a manual loop.

**Break:**
1. Handle empty list.
2. Assume first element is largest.
3. Loop through the rest and compare.

**Logic:**
```
largest = nums[0]
for n in nums: if n > largest: largest = n
return largest
```


````python
def largest(nums):
    if not nums:
        return None
    big = nums[0]
    for n in nums[1:]:
        if n > big:
            big = n
    return big

print(largest([12, 45, 7, 89, 23]))   # 89
print(largest([-5, -2, -9]))           # -2
print(largest([]))                     # None
````

## Q4. Find the First Longest Word When Ties Occur
**Observe:** Input: sentence; output: one word. `Python makes coding useful` → `Python`.

**Understand:** Return the longest word; on a tie, the earliest one.

**Break:**
1. Split into words.
2. Track the longest so far.
3. Replace only if strictly longer.

**Logic:**
```
longest = words[0]
for w in words: if len(w) > len(longest): longest = w
```
Using strict `>` means earlier words win ties.


````python
def first_longest(sentence):
    words = sentence.split()
    if not words:
        return None
    longest = words[0]
    for w in words[1:]:
        if len(w) > len(longest):
            longest = w
    return longest

print(first_longest("Python makes coding useful"))   # Python
print(first_longest("a bb ccc dd"))                   # ccc
print(first_longest(""))                              # None
````

## Q5. Find the Word with the Maximum Number of Characters
**Observe:** Input: sentence; output: word. `Data engineering requires problem solving` → `engineering`.

**Understand:** Find the word with the highest character count.

**Break:**
1. Split into words.
2. Track the word with the greatest length.

**Logic:**
```
longest = words[0]; for w in words: if len(w) > len(longest): longest = w
```


````python
def max_char_word(sentence):
    words = sentence.split()
    if not words:
        return None
    best = words[0]
    for w in words:
        if len(w) > len(best):
            best = w
    return best

print(max_char_word("Data engineering requires problem solving"))   # engineering
print(max_char_word("sun moon stars"))                               # stars
print(max_char_word("hi"))                                           # hi
````

## Q6. Count Vowels in a String

## Q7. Count Consonants in a String
**Observe:** Input: string; output: integer. `Python 123!` → `5`.

**Understand:** Count only alphabetic letters that are not vowels; ignore spaces, digits, symbols.

**Break:**
1. Loop through characters.
2. Check `isalpha()`.
3. Check not a vowel.
4. Count.

**Logic:**
```
if ch.isalpha() and ch.lower() not in 'aeiou': count += 1
```


````python
def count_consonants(s):
    count = 0
    for ch in s:
        if ch.isalpha() and ch.lower() not in "aeiou":
            count += 1
    return count

print(count_consonants("Python 123!"))   # 5
print(count_consonants("AEIOU"))          # 0
print(count_consonants("Hello, World!"))  # 7
````

## Q8. Find the Most Frequent Character
**Observe:** Input: string; output: one character. `banana` → `a`.

**Understand:** Find the character with the highest count (ignoring spaces); ties go to the first encountered.

**Break:**
1. Remove spaces.
2. Count each character in a dictionary (keeps first-appearance order).
3. Pick the highest count, first one on ties.

**Logic:**
```
counts = {}
for ch in s (not space): counts[ch] += 1
best = first key; for ch in counts: if counts[ch] > counts[best]: best = ch
```


````python
def most_frequent_char(s):
    counts = {}
    for ch in s:
        if ch != " ":
            counts[ch] = counts.get(ch, 0) + 1
    if not counts:
        return None
    best = next(iter(counts))            # first encountered
    for ch in counts:
        if counts[ch] > counts[best]:    # strict '>' keeps first on ties
            best = ch
    return best

print(most_frequent_char("banana"))      # a
print(most_frequent_char("abcab c"))     # a  (a=2,b=2,c=2 -> first)
print(most_frequent_char("   "))         # None
````

## Q10. Find All Characters Appearing More Than Once
**Observe:** Input: string; output: list. `programming` → `['r', 'g', 'm']`.

**Understand:** List characters with count > 1, each once, in first-appearance order, ignoring spaces.

**Break:**
1. Count characters (dictionary preserves order).
2. Collect those with count > 1.

**Logic:**
```
counts = {...}
return [ch for ch in counts if counts[ch] > 1]
```


````python
def repeated_chars(s):
    counts = {}
    for ch in s:
        if ch != " ":
            counts[ch] = counts.get(ch, 0) + 1
    return [ch for ch in counts if counts[ch] > 1]

print(repeated_chars("programming"))   # ['r', 'g', 'm']
print(repeated_chars("abc"))            # []
print(repeated_chars("a b a b"))        # ['a', 'b']
````

## Q16. Find the Most Frequent Word in a Sentence
**Observe:** Input: sentence; output: lowercase word. `Python is easy and Python is powerful` → `python`.

**Understand:** Case-insensitive word counting; earliest word wins ties.

**Break:**
1. Lowercase and split.
2. Count in an ordered dictionary.
3. Pick the highest count, first on ties.

**Logic:**
```
words = sentence.lower().split()
counts = {}; best = first key; strict '>' comparison
```


````python
def most_frequent_word(sentence):
    counts = {}
    for w in sentence.lower().split():
        counts[w] = counts.get(w, 0) + 1
    if not counts:
        return None
    best = next(iter(counts))
    for w in counts:
        if counts[w] > counts[best]:
            best = w
    return best

print(most_frequent_word("Python is easy and Python is powerful"))   # python
print(most_frequent_word("Red blue RED Blue green"))                  # red
print(most_frequent_word(""))                                         # None
````

## Q18. Find the First Word That Appears Only Once
**Observe:** Input: sentence; output: word. `apple banana apple mango banana orange` → `mango`.

**Understand:** First word (in sentence order) whose count is exactly 1, case-insensitive.

**Break:**
1. Lowercase + split.
2. Count all words.
3. Scan in original order and return the first with count 1.

**Logic:**
```
for w in words: if counts[w] == 1: return w
return None
```


````python
def first_unique_word(sentence):
    words = sentence.lower().split()
    counts = {}
    for w in words:
        counts[w] = counts.get(w, 0) + 1
    for w in words:
        if counts[w] == 1:
            return w
    return None

print(first_unique_word("apple banana apple mango banana orange"))   # mango
print(first_unique_word("a A b B"))                                   # None
print(first_unique_word("Only"))                                      # only
````

## Q20. Find the Longest Word That Appears Only Once
**Observe:** Input: sentence; output: word. `Python makes coding useful Python` → `coding`.

**Understand:** Among words occurring exactly once (case-insensitive), return the longest; first on ties.

**Break:**
1. Lowercase + split.
2. Count words.
3. Scan in order, consider only count == 1 words, keep the longest (strict `>`).

**Logic:**
```
best = None
for w in words: if counts[w]==1 and (best is None or len(w) > len(best)): best = w
```


````python
def longest_unique_word(sentence):
    words = sentence.lower().split()
    counts = {}
    for w in words:
        counts[w] = counts.get(w, 0) + 1
    best = None
    for w in words:
        if counts[w] == 1 and (best is None or len(w) > len(best)):
            best = w
    return best

print(longest_unique_word("Python makes coding useful Python"))   # coding
print(longest_unique_word("go go GO"))                            # None
print(longest_unique_word("cat bird dog fish"))                   # bird
````
