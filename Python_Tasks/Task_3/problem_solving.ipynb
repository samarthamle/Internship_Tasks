## Q1. Find the Frequency of Each Word in a Sentence

### Pseudocode
1. Create an empty dictionary.
2. Split the sentence into words.
3. For each word, increase its count in the dictionary.
4. Return the dictionary.

### Expected Output
```text
{'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}
{'data': 2, 'science': 1}
```

````python
def word_frequency(sentence):
    frequency = {}

    for word in sentence.split():
        frequency[word] = frequency.get(word, 0) + 1

    return frequency


# Original example
print(word_frequency("python is easy and python is powerful"))

# Additional test case
print(word_frequency("data data science"))
````

## Q2. Find the Frequency of Each Character in a Word

### Pseudocode
1. Create an empty dictionary.
2. Visit each character in the word.
3. Increase that character's count.
4. Return the dictionary.

### Expected Output
```text
{'b': 1, 'a': 3, 'n': 2}
{'h': 1, 'e': 1, 'l': 2, 'o': 1}
```

````python
def character_frequency(word):
    frequency = {}

    for char in word:
        frequency[char] = frequency.get(char, 0) + 1

    return frequency


# Original example
print(character_frequency("banana"))

# Additional test case
print(character_frequency("hello"))
````

## Q3. Find the First Non-Repeating Character in a Word

### Pseudocode
1. Count the frequency of every character.
2. Scan the word from left to right.
3. Return the first character whose frequency is 1.
4. If none exists, return None.

### Expected Output
```text
w
None
```

````python
def first_non_repeating_character(word):
    frequency = character_frequency(word)

    for char in word:
        if frequency[char] == 1:
            return char

    return None


# Original example
print(first_non_repeating_character("swiss"))

# Additional test case
print(first_non_repeating_character("aabb"))
````

## Q4. Check Whether Two Strings Are Anagrams

### Pseudocode
1. Remove spaces and make both strings lowercase.
2. Count the characters in each string.
3. Compare the two frequency dictionaries.
4. Return True if they match; otherwise return False.

### Expected Output
```text
True
False
```

````python
def are_anagrams(s1, s2):
    s1 = s1.replace(" ", "").lower()
    s2 = s2.replace(" ", "").lower()

    return character_frequency(s1) == character_frequency(s2)


# Original example
print(are_anagrams("listen", "silent"))

# Additional test case
print(are_anagrams("hello", "world"))
````

## Q5. Find the First Repeating Character in a String

### Pseudocode
1. Create an empty set to remember characters already seen.
2. Scan the string from left to right.
3. If a character is already in the set, return it.
4. Otherwise add it to the set.
5. If no character repeats, return None.

### Expected Output
```text
r
None
```

````python
def first_repeating_character(text):
    seen = set()

    for char in text:
        if char in seen:
            return char
        seen.add(char)

    return None


# Original example
print(first_repeating_character("programming"))

# Additional test case
print(first_repeating_character("abcdef"))
````

## Q6. Move All Zeros to the End of a List

### Pseudocode
1. Collect all non-zero numbers in their original order.
2. Collect all zero values.
3. Join the non-zero list and zero list.
4. Return the new list.

### Expected Output
```text
[3, 1, 2, 4, 0, 0, 0]
[1, 2, 0]
```

````python
def move_zeros_to_end(nums):
    non_zeros = [num for num in nums if num != 0]
    zeros = [num for num in nums if num == 0]

    return non_zeros + zeros


# Original example
print(move_zeros_to_end([0, 3, 0, 1, 2, 0, 4]))

# Additional test case
print(move_zeros_to_end([1, 0, 2]))
````

## Q7. Find the Second Highest Distinct Number in a List

### Pseudocode
1. Remove duplicate numbers.
2. Sort the unique numbers from largest to smallest.
3. If fewer than two unique numbers exist, return None.
4. Otherwise return the second number.

### Expected Output
```text
30
None
```

````python
def second_highest_distinct(nums):
    unique_nums = sorted(set(nums), reverse=True)

    if len(unique_nums) < 2:
        return None

    return unique_nums[1]


# Original example
print(second_highest_distinct([10, 20, 40, 30, 40, 20]))

# Additional test case
print(second_highest_distinct([5, 5]))
````

## Q8. Remove Duplicates Without Changing the Original Order

### Pseudocode
1. Create an empty result list.
2. Visit each value in the input list.
3. If the value is not already in the result, append it.
4. Return the result list.

### Expected Output
```text
[4, 2, 5, 7]
[1, 2, 3]
```

````python
def remove_duplicates(nums):
    result = []

    for num in nums:
        if num not in result:
            result.append(num)

    return result


# Original example
print(remove_duplicates([4, 2, 4, 5, 2, 7, 5]))

# Additional test case
print(remove_duplicates([1, 1, 2, 3, 2]))
````

## Q9. Find the Missing Number from 1 to N

### Pseudocode
1. Calculate the sum of all numbers from 1 to N using N × (N + 1) // 2.
2. Calculate the sum of the numbers in the list.
3. Subtract the list sum from the expected sum.
4. Return the missing number.

### Expected Output
```text
3
2
```

````python
def find_missing_number(nums, n):
    expected_sum = n * (n + 1) // 2
    return expected_sum - sum(nums)


# Original example
print(find_missing_number([1, 2, 4, 5, 6], 6))

# Additional test case
print(find_missing_number([1, 3, 4], 4))
````

## Q10. Find Two Numbers Whose Sum Equals a Target

### Pseudocode
1. Create a set of previously seen numbers.
2. Scan numbers from left to right.
3. For each number, calculate target minus that number.
4. If the needed number has already been seen, return the pair in input order.
5. Otherwise save the current number. Return an empty list if no pair exists.

### Expected Output
```text
[2, 7]
[4, 6]
```

````python
def two_sum_first_pair(nums, target):
    seen = set()

    for num in nums:
        needed = target - num

        if needed in seen:
            return [needed, num]

        seen.add(num)

    return []


# Original example
print(two_sum_first_pair([2, 7, 11, 15], 9))

# Additional test case
print(two_sum_first_pair([1, 4, 6, 8], 10))
````

## Q11. Find the Number Closest to Zero

### Pseudocode
1. If the list is empty, return None.
2. Compare numbers using their absolute values.
3. If two numbers have the same distance from zero, prefer the positive number.
4. Return the selected number.

### Expected Output
```text
2
-1
```

````python
def closest_to_zero(nums):
    if not nums:
        return None

    return min(nums, key=lambda num: (abs(num), -num))


# Original example
print(closest_to_zero([-8, -3, 2, 5, -2]))

# Additional test case
print(closest_to_zero([-4, 4, -1]))
````

## Q12. Find Common Elements Between Two Lists

### Pseudocode
1. Create an empty result list.
2. Scan the first list in its original order.
3. If an item is in the second list and not already in the result, append it.
4. Return the result.

### Expected Output
```text
[3, 4, 5]
[2, 4]
```

````python
def common_elements(list1, list2):
    result = []

    for item in list1:
        if item in list2 and item not in result:
            result.append(item)

    return result


# Original example
print(common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))

# Additional test case
print(common_elements([2, 2, 3, 4], [4, 2]))
````

## Q13. Rotate a List to the Right by K Positions

### Pseudocode
1. If the list is empty, return an empty list.
2. Reduce K using K modulo the list length.
3. Take the last K elements and place them first.
4. Follow them with the remaining elements.

### Expected Output
```text
[4, 5, 1, 2, 3]
[3, 1, 2]
```

````python
def rotate_right(nums, k):
    if not nums:
        return []

    k = k % len(nums)

    if k == 0:
        return nums.copy()

    return nums[-k:] + nums[:-k]


# Original example
print(rotate_right([1, 2, 3, 4, 5], 2))

# Additional test case
print(rotate_right([1, 2, 3], 4))
````

## Q14. Find the Maximum Sum of Two Consecutive Elements

### Pseudocode
1. Add each element to the element immediately after it.
2. Compare all consecutive-pair sums.
3. Return the largest sum.

### Expected Output
```text
14
-7
```

````python
def maximum_consecutive_pair_sum(nums):
    return max(nums[i] + nums[i + 1] for i in range(len(nums) - 1))


# Original example
print(maximum_consecutive_pair_sum([4, 7, 2, 9, 5]))

# Additional test case
print(maximum_consecutive_pair_sum([-5, -2, -7]))
````

## Q15. Find All Unique Pairs Whose Sum Equals a Target

### Pseudocode
1. Check every pair of different positions in the list.
2. If the two values add to the target, arrange the pair in ascending order.
3. Store pairs in a set so duplicates are removed.
4. Sort the pairs and return them as lists.

### Expected Output
```text
[[1, 8], [2, 7], [4, 5]]
[[1, 5], [2, 4]]
```

````python
def all_pairs_with_target_sum(nums, target):
    pairs = set()

    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                pair = tuple(sorted([nums[i], nums[j]]))
                pairs.add(pair)

    return [list(pair) for pair in sorted(pairs)]


# Original example
print(all_pairs_with_target_sum([2, 4, 3, 5, 7, 8, 1], 9))

# Additional test case
print(all_pairs_with_target_sum([1, 2, 3, 4, 5], 6))
````
